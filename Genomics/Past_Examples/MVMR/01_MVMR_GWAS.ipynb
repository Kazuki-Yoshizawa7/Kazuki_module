```python
def standardize_gwas_data(df, col_mapping):
    """
    GWASデータを標準フォーマットに変換
    
    標準フォーマット:
    - SNP: RS ID
    - chr: 染色体
    - pos: 位置
    - effect_allele: 効果アリル
    - other_allele: 他方のアリル
    - eaf: 効果アリル頻度
    - beta: 効果量
    - se: 標準誤差
    - pval: P値
    """
    df_std = df.copy()
    
    # カラム名を変更
    df_std = df_std.rename(columns=col_mapping)
    
    # 必要なカラムのみ選択
    required_cols = ['SNP', 'chr', 'pos', 'effect_allele', 'other_allele', 
                     'eaf', 'beta', 'se', 'pval']
    
    # 存在しないカラムはNAで埋める
    for col in required_cols:
        if col not in df_std.columns:
            df_std[col] = np.nan
    
    df_std = df_std[required_cols]
    
    # アリルを大文字に統一
    df_std['effect_allele'] = df_std['effect_allele'].str.upper()
    df_std['other_allele'] = df_std['other_allele'].str.upper()
    
    # 数値型に変換
    numeric_cols = ['chr', 'pos', 'eaf', 'beta', 'se', 'pval']
    for col in numeric_cols:
        df_std[col] = pd.to_numeric(df_std[col], errors='coerce')
    
    return df_std

# 2-1. Snoreデータの標準化
snore_mapping = {
    'SNP': 'SNP',
    'CHR': 'chr',
    'BP': 'pos',
    'A1': 'effect_allele',
    'A2': 'other_allele',
    'FREQ': 'eaf',
    'BETA': 'beta',
    'SE': 'se',
    'P': 'pval'
}
snore_std = standardize_gwas_data(df_snore, snore_mapping)

# 2-2. Neckデータの標準化
neck_mapping = {
    'variant_id': 'SNP',
    'chromosome': 'chr',
    'base_pair_location': 'pos',
    'effect_allele': 'effect_allele',
    'other_allele': 'other_allele',
    'effect_allele_frequency': 'eaf',
    'beta': 'beta',
    'standard_error': 'se',
    'p_value': 'pval'
}
neck_std = standardize_gwas_data(df_neck, neck_mapping)

# 2-3. BMIデータの標準化
bmi_mapping = {
    'variant_id': 'SNP',
    'chromosome':'chr',
    'effect_allele': 'effect_allele',
    'other_allele': 'other_allele',
    'effect_allele_frequency': 'eaf',
    'beta': 'beta',
    'standard_error': 'se',
    'p_value': 'pval'
}
bmi_std = standardize_gwas_data(df_bmi, bmi_mapping)

print("Snore data shape:", snore_std.shape)
print("Neck data shape:", neck_std.shape)
print("BMI data shape:", bmi_std.shape)
```

```python
def clean_gwas_data(df, pval_threshold=1.0):
    """
    GWASデータのクリーニング
    - 欠損値の除去
    - 無効な値の除去
    - P値でフィルタリング
    """
    df_clean = df.copy()
    
    # 必須カラムの欠損値を除去
    essential_cols = ['SNP', 'effect_allele', 'other_allele', 'beta', 'se', 'pval']
    df_clean = df_clean.dropna(subset=essential_cols)
    
    # P値でフィルタリング
    df_clean = df_clean[df_clean['pval'] <= pval_threshold]
    
    # SEが0または負の値を除去
    df_clean = df_clean[df_clean['se'] > 0]
    
    # 重複SNPを除去（最も低いP値を保持）
    df_clean = df_clean.sort_values('pval').drop_duplicates(subset=['SNP'], keep='first')
    
    return df_clean

snore_clean = clean_gwas_data(snore_std)
neck_clean = clean_gwas_data(neck_std, pval_threshold=1e-5)  # 有意なSNPのみ
bmi_clean = clean_gwas_data(bmi_std, pval_threshold=5e-8)   # 有意なSNPのみ

print("\nAfter cleaning:")
print("Snore data shape:", snore_clean.shape)
print("Neck significant SNPs:", neck_clean.shape)
print("BMI significant SNPs:", bmi_clean.shape)
```

```python
def harmonize_alleles(exp_df, out_df):
    """
    曝露とアウトカムのアリルの向きを揃える
    """
    # SNPでマージ
    merged = pd.merge(
        exp_df, 
        out_df, 
        on='SNP', 
        suffixes=('_exp', '_out'),
        how='inner'
    )
    
    # アリルが一致するケースと反転するケースを判定
    match = (
        (merged['effect_allele_exp'] == merged['effect_allele_out']) &
        (merged['other_allele_exp'] == merged['other_allele_out'])
    )
    
    flip = (
        (merged['effect_allele_exp'] == merged['other_allele_out']) &
        (merged['other_allele_exp'] == merged['effect_allele_out'])
    )
    
    # 反転が必要なケースでbetaとeafを反転
    merged.loc[flip, 'beta_out'] = -merged.loc[flip, 'beta_out']
    merged.loc[flip, 'eaf_out'] = 1 - merged.loc[flip, 'eaf_out']
    
    # アリルが一致しないケースを除去（アンビギュアスSNPなど）
    merged = merged[match | flip]
    
    print(f"  Matched SNPs: {match.sum()}")
    print(f"  Flipped SNPs: {flip.sum()}")
    print(f"  Total harmonized: {len(merged)}")
    
    return merged

# ============================================
# 5. 各曝露とアウトカムをハーモナイズ
# ============================================

print("\nHarmonizing Neck with Snore:")
neck_harm = harmonize_alleles(neck_clean, snore_clean)

print("\nHarmonizing BMI with Snore:")
bmi_harm = harmonize_alleles(bmi_clean, snore_clean)
```

```python
# オプション1: 両方の曝露で共通のSNPを使う（厳格）
common_snps_strict = set(neck_harm['SNP']) & set(bmi_harm['SNP'])
print(f"\nCommon SNPs across all exposures (strict): {len(common_snps_strict)}")

# オプション2: どちらかの曝露で有意なら含める（推奨）
# MVMRでは各曝露に特異的なSNPも使える
all_snps = set(neck_harm['SNP']) | set(bmi_harm['SNP'])
print(f"Total unique SNPs (union): {len(all_snps)}")
```

```python
common_snps = common_snps_strict
neck_final = neck_harm[neck_harm['SNP'].isin(common_snps)].sort_values('SNP').reset_index(drop=True)
bmi_final = bmi_harm[bmi_harm['SNP'].isin(common_snps)].sort_values('SNP').reset_index(drop=True)
    
mvmr_data = pd.DataFrame({
        'SNP': neck_final['SNP'],
        'chr': neck_final['chr_exp'],
        'pos': neck_final['pos_exp'],
        'effect_allele': neck_final['effect_allele_exp'],
        'other_allele': neck_final['other_allele_exp'],
        
        'beta_neck': neck_final['beta_exp'],
        'se_neck': neck_final['se_exp'],
        'pval_neck': neck_final['pval_exp'],
        'eaf_neck': neck_final['eaf_exp'],
        
        'beta_bmi': bmi_final['beta_exp'],
        'se_bmi': bmi_final['se_exp'],
        'pval_bmi': bmi_final['pval_exp'],
        'eaf_bmi': bmi_final['eaf_exp'],
        
        'beta_snore': neck_final['beta_out'],
        'se_snore': neck_final['se_out'],
        'pval_snore': neck_final['pval_out'],
        'eaf_snore': neck_final['eaf_out']
})

print("\nFinal MVMR dataset shape:", mvmr_data.shape)
print("\nFirst few rows:")
print(mvmr_data.head())
```

```python

# ============================================
# Union版の作成（こちらを追加）
# ============================================



all_snps_union = set(neck_harm['SNP']) | set(bmi_harm['SNP'])
print(f"Number of SNPs: {len(all_snps_union)}")

# Neckのデータ準備
neck_for_merge = neck_harm[['SNP', 'chr_exp', 'pos_exp', 'effect_allele_exp', 
                              'other_allele_exp', 'beta_exp', 'se_exp', 
                              'pval_exp', 'eaf_exp', 'beta_out', 'se_out', 
                              'pval_out', 'eaf_out']].copy()
neck_for_merge.columns = ['SNP', 'chr', 'pos', 'effect_allele', 'other_allele',
                            'beta_neck', 'se_neck', 'pval_neck', 'eaf_neck',
                            'beta_snore_neck', 'se_snore_neck', 'pval_snore_neck', 'eaf_snore_neck']

# BMIのデータ準備
bmi_for_merge = bmi_harm[['SNP', 'chr_exp', 'pos_exp', 'effect_allele_exp',
                            'other_allele_exp', 'beta_exp', 'se_exp',
                            'pval_exp', 'eaf_exp', 'beta_out', 'se_out',
                            'pval_out', 'eaf_out']].copy()
bmi_for_merge.columns = ['SNP', 'chr', 'pos', 'effect_allele', 'other_allele',
                          'beta_bmi', 'se_bmi', 'pval_bmi', 'eaf_bmi',
                          'beta_snore_bmi', 'se_snore_bmi', 'pval_snore_bmi', 'eaf_snore_bmi']

# Outer join
mvmr_data_union = pd.merge(neck_for_merge, bmi_for_merge, 
                     on=['SNP'],
                     how='outer',
                     suffixes=('_from_neck', '_from_bmi'))

# chr, pos, alleleの情報を統合
mvmr_data_union['chr'] = mvmr_data_union['chr_from_neck'].fillna(mvmr_data_union['chr_from_bmi'])
mvmr_data_union['pos'] = mvmr_data_union['pos_from_neck'].fillna(mvmr_data_union['pos_from_bmi'])
mvmr_data_union['effect_allele'] = mvmr_data_union['effect_allele_from_neck'].fillna(mvmr_data_union['effect_allele_from_bmi'])
mvmr_data_union['other_allele'] = mvmr_data_union['other_allele_from_neck'].fillna(mvmr_data_union['other_allele_from_bmi'])

# Snoreのbeta/seを統合
mvmr_data_union['beta_snore'] = mvmr_data_union['beta_snore_bmi'].fillna(mvmr_data_union['beta_snore_neck'])
mvmr_data_union['se_snore'] = mvmr_data_union['se_snore_bmi'].fillna(mvmr_data_union['se_snore_neck'])
mvmr_data_union['pval_snore'] = mvmr_data_union['pval_snore_bmi'].fillna(mvmr_data_union['pval_snore_neck'])
mvmr_data_union['eaf_snore'] = mvmr_data_union['eaf_snore_bmi'].fillna(mvmr_data_union['eaf_snore_neck'])

# 曝露のbeta/seで欠損値を0と大きなSEで埋める
mvmr_data_union['beta_neck'] = mvmr_data_union['beta_neck'].fillna(0)
mvmr_data_union['se_neck'] = mvmr_data_union['se_neck'].fillna(1.0)
mvmr_data_union['pval_neck'] = mvmr_data_union['pval_neck'].fillna(1.0)
mvmr_data_union['eaf_neck'] = mvmr_data_union['eaf_neck'].fillna(0.5)

mvmr_data_union['beta_bmi'] = mvmr_data_union['beta_bmi'].fillna(0)
mvmr_data_union['se_bmi'] = mvmr_data_union['se_bmi'].fillna(1.0)
mvmr_data_union['pval_bmi'] = mvmr_data_union['pval_bmi'].fillna(1.0)
mvmr_data_union['eaf_bmi'] = mvmr_data_union['eaf_bmi'].fillna(0.5)

# 最終的なカラムのみ選択
mvmr_data_union = mvmr_data_union[[
    'SNP', 'chr', 'pos', 'effect_allele', 'other_allele',
    'beta_neck', 'se_neck', 'pval_neck', 'eaf_neck',
    'beta_bmi', 'se_bmi', 'pval_bmi', 'eaf_bmi',
    'beta_snore', 'se_snore', 'pval_snore', 'eaf_snore'
]].sort_values('SNP').reset_index(drop=True)

# 保存
mvmr_data_union.to_csv("mvmr_union_all_snps.csv", index=False)
```
