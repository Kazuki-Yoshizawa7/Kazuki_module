```python
from scipy.stats import gaussian_kde
from scipy.optimize import brentq
import numpy as np

# ダミーデータ
data_P = np.array([1, 1.5, 2, 2.5])
data_Q = np.array([4, 4.5, 5, 5.5])

# 1. KDE関数の定義
kde_P = gaussian_kde(data_P)
kde_Q = gaussian_kde(data_Q)

# 2. 差分関数の定義
def difference_function(x):
    """P(x) - Q(x) を計算する関数"""
    return kde_P(x) - kde_Q(x)

# 3. 交点が存在する区間を特定 (例: 2.5 から 4.5 の間にあると推定)
# 注: brentq を使うには、区間の両端で関数の符号が異なっている必要があります
try:
    # 4. 求根アルゴリズムで正確な交点 X を計算
    # 探索範囲 [a, b] を指定
    intersection_x = brentq(difference_function, 2.5, 4.5)
    
    # 交点でのY値も取得
    intersection_y = kde_P(intersection_x)[0]

    print(f"✅ 数値計算による正確な交点座標:")
    print(f"X = {intersection_x:.4f}, Y = {intersection_y:.4f}")

except ValueError:
    print("❌ 指定された区間で符号の変化が見られませんでした。探索範囲を見直してください。")
```
