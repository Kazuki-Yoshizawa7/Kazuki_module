```python
import re 
import time 
import pandas as pd
import numpy as np
```

```python
def main():
    base_url = "https://www.intcul.tohoku.ac.jp"
    list_url = f"{base_url}/ja/news/"

    # ユーザーエージェントを設定（アクセス拒否を防ぐためのお守り）
    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36"
    }

    # --- ステップ1: 一覧ページから記事のURLを集める ---
    print(f"一覧ページを取得中: {list_url}")
    response = requests.get(list_url, headers=headers)
    response.encoding = response.apparent_encoding

    article_urls = []

    if response.status_code == 200:
        soup = BeautifulSoup(response.text, 'html.parser')
        
        # ページ内のすべてのリンク(aタグ)を取得
        all_links = soup.find_all('a')
        
        for link in all_links:
            href = link.get('href')
            if href:
                # "/ja/news/article/" を含み、かつ末尾が数字であるURLを探す
                # URLが相対パスの場合もあるので絶対パスに変換してチェック
                if href.startswith('/'):
                    full_url = base_url + href
                else:
                    full_url = href
                
                # 正規表現でチェック: .../article/数字
                if re.search(r'/ja/news/article/\d+$', full_url):
                    article_urls.append(full_url)
        
        # 重複を除去してリスト化
        article_urls = sorted(list(set(article_urls)), reverse=True)
        print(f"{len(article_urls)} 件の記事URLが見つかりました。")
    else:
        print("一覧ページの取得に失敗しました")
        return

    # --- ステップ2: 各記事ページから情報を抽出する ---
    data_list = []
    
    for url in article_urls:
        print(f"スクレイピング中: {url}")
        
        try:
            # サーバー負荷軽減のため1秒待機
            time.sleep(1)
            
            res_article = requests.get(url, headers=headers)
            res_article.encoding = res_article.apparent_encoding
            
            if res_article.status_code == 200:
                soup_article = BeautifulSoup(res_article.text, 'html.parser')
                
                # --- 指定の抽出ロジック ---
                
                # 1. Article番号をURLから抽出
                # URLの末尾(数字部分)を取得
                article_id = url.split('/')[-1]
                
                # 2. h1タグの2つ目 (インデックス1) を取得
                h1_tags = soup_article.select("h1")
                
                if len(h1_tags) >= 2:
                    title = h1_tags[1].text.strip()
                else:
                    # h1が2つ無い場合のフォールバック（1つ目を取得など）
                    title = h1_tags[0].text.strip() if h1_tags else "タイトルなし"
                
                # データをリストに追加
                data_list.append({
                    'Article番号': article_id,
                    'タイトル': title,
                    'URL': url
                })
                
            else:
                print(f"記事取得失敗: {res_article.status_code}")
                
        except Exception as e:
            print(f"エラー発生: {e}")
            continue

    # --- ステップ3: Pandasで保存 ---
    if data_list:
        df = pd.DataFrame(data_list)
        
        # 結果を表示
        print("\n--- 取得結果 ---")
        print(df.head())
        
        # CSV保存
        df.to_csv('tohoku_news_articles.csv', index=False, encoding='utf-8-sig')
        print("\n'tohoku_news_articles.csv' に保存しました。")
    else:
        print("データが取得できませんでした。")

if __name__ == "__main__":
    main()
```
