#### 半導体指数やARMの株価がソフトバンクGの株価にどれくらい影響を与えているか？　ラグ成分などをとって調べてみる。特にアメリカは1日遅れていることを生かしてその次の日の日経に反映されているという認識

```python
ticker_list = {

    "^SOX",
    "ARM",
    "9984.T"

}

plt.figure(figsize=(20,10))


try:
    for i in ticker_list:
        data = yf.download(i,start="2019-01-01", end="2026-01-31")
        dt = data.reset_index()
        scaler = StandardScaler()

        dt['Close_z'] = scaler.fit_transform(
            dt[['Close']]
        )
        plt.plot(dt['Date'],dt["Close_z"], label=f"Code:{i}")
except Exception as e:
    print(e)

plt.legend()
plt.grid()
plt.show()
```

```python


tickers = {
    "SOX": "^SOX",
    "ARM": "ARM",
    "SBG": "9984.T"
}

dfs = []

for name, ticker in tickers.items():
    df = yf.download(
        ticker,
        start="2019-01-01",
        end="2026-01-031",
        auto_adjust=False,
        progress=False
    )[['Close']]
    df.columns = [name]
    dfs.append(df)

# 日付で結合
price = pd.concat(dfs, axis=1).dropna()
price_norm = price / price.iloc[0]
ret = np.log(price).diff().dropna()
ret.corr()
rolling_corr = ret['SBG'].rolling(60).corr(ret['ARM'])

rolling_corr.plot(
    title="Rolling Correlation: SBG vs ARM (60d)",
    figsize=(10,5)
)
```

```python
lag_df = ret.copy()
lag_df['ARM_lag1'] = lag_df['ARM'].shift(1) # ARM の前日の変動がどの程度翌日のSoftbankのデータに聞いているかを見ている
lag_df['SOX_lag1'] = lag_df['SOX'].shift(1) # 前日の（半導体指数）が与える影響

lag_df = lag_df.dropna()

import statsmodels.api as sm

X = lag_df[['ARM_lag1', 'SOX_lag1']]
X = sm.add_constant(X)
y = lag_df['SBG']

model = sm.OLS(y, X).fit()
print(model.summary())
```

```python
beta = model.params
contrib = X.mul(beta, axis=1)

contrib[['ARM_lag1', 'SOX_lag1']].cumsum().plot(
    figsize=(12,6),
    title="Cumulative Contribution to SBG Return"
)
```
