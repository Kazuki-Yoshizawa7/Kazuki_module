```python
import pandas as pd
import yfinance as yf 
import numpy as np 
import matplotlib.pyplot as plt
import datetime
import seaborn as sns 
from kazustat import YFinanceAnalysis
import warnings 
from sklearn.preprocessing import StandardScaler
```

チャート分析

```python

# Tickers (Tokyo Stock Exchange)
tickers = {
    "Fujifilm": "4901.T",
    "Nippon Steel": "5401.T",
    "Shin-Etsu Chemical": "4063.T",
}

# Download 1 year of price data
data = yf.download(list(tickers.values()), period="2y", auto_adjust=True)["Close"]
data = data.rename(columns={v: k for k, v in tickers.items()})  # maps by value — correct

# Normalized price (base = 100 at start)
normalized = data / data.iloc[0] * 100

# Plot
fig, axes = plt.subplots(2, 1, figsize=(12, 10))

# Raw closing price
data.plot(ax=axes[0], title="Closing Price (JPY)")
axes[0].set_ylabel("Price (JPY)")
axes[0].legend()
axes[0].grid(True)

# Normalized (relative performance)
normalized.plot(ax=axes[1], title="Normalized Price (Base = 100)")
axes[1].set_ylabel("Indexed Price")
axes[1].axhline(100, color="gray", linestyle="--", linewidth=0.8)
axes[1].legend()
axes[1].grid(True)

plt.tight_layout()
plt.show()
```
